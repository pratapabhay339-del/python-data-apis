## 1. Load the dataset

```python
# In Colab, upload student_performance.csv first.
df = pd.read_csv('student_performance.csv')
df.head()
```

## 2. Inspect and clean data

```python
print(df.info())
print(df.isna().sum())
df = df.drop_duplicates().copy()
numeric_cols = ['attendance_percent','study_hours','assignments_completed','midterm_score','final_score']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
    df[col] = df[col].fillna(df[col].median())
df['branch'] = df['branch'].str.strip().str.upper()
df.describe()
```

## 3. Analyze

```python
print('Average final score:', np.mean(df['final_score']))
print('Pass rate:', (df['final_score'] >= 40).mean() * 100, '%')
branch_summary = df.groupby('branch').agg(
    students=('student_id','count'),
    average_final_score=('final_score','mean'),
    average_attendance=('attendance_percent','mean')
).round(2)
branch_summary
```

## 4. Visualize

```python
sns.histplot(df['final_score'], bins=6, kde=True)
plt.title('Distribution of Final Scores')
plt.show()
```

```python
sns.heatmap(df.select_dtypes('number').corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

## 5. Consume a REST API and process JSON

```python
url = 'https://jsonplaceholder.typicode.com/posts'
response = requests.get(url, timeout=10)
response.raise_for_status()
payload = response.json()
api_df = pd.json_normalize(payload)
api_df.head()
```
